# Web Scraping — Dados do Oscar (2010–2015)

Este notebook extrai os dados da página **Scrape This Site — AJAX and Javascript**.





## Etapa 1 — Instalar e importar as bibliotecas

- **requests**: realiza as requisições ao site;
- **pandas**: organiza os dados em tabela;
- **time**: cria uma pequena pausa entre as requisições;
- **google.colab.files**: permite baixar o Excel ao final.


```python
!pip -q install openpyxl

import requests
import pandas as pd
import time

from google.colab import files

print("Bibliotecas carregadas com sucesso!")
```

## Etapa 2 — Definir a página e os anos

O exercício possui dados de **2010 a 2015**.

Será criada uma lista chamada `todos_filmes` para armazenar os registros extraídos de todos os anos.


```python
url = "https://www.scrapethissite.com/pages/ajax-javascript/"

anos = range(2010, 2016)

todos_filmes = []

print("Anos que serão consultados:", list(anos))
```

## Etapa 3 — Extrair os dados de cada ano

Para cada ano, o código envia os parâmetros:

- `ajax=true`
- `year=ANO`

A resposta é convertida de JSON para objetos Python. Em seguida, cada filme é adicionado à lista geral.


```python
for ano in anos:

    print(f"Extraindo dados do Oscar de {ano}...")

    parametros = {
        "ajax": "true",
        "year": ano
    }

    resposta = requests.get(
        url,
        params=parametros,
        timeout=30
    )

    # Interrompe a execução se houver erro HTTP
    resposta.raise_for_status()

    # Converte a resposta JSON para uma estrutura Python
    filmes = resposta.json()

    for filme in filmes:

        todos_filmes.append({
            "Ano": filme.get("year", ano),
            "Filme": filme.get("title"),
            "Indicações": filme.get("nominations"),
            "Prêmios": filme.get("awards"),
            "Melhor Filme": filme.get("best_picture", False)
        })

    print(f"✓ {len(filmes)} filmes encontrados em {ano}")

    # Pequena pausa entre as consultas
    time.sleep(0.5)

print("\nExtração concluída!")
print("Total de registros extraídos:", len(todos_filmes))
```

## Etapa 4 — Criar uma única tabela

Agora os registros de todos os anos serão transformados em um `DataFrame` do pandas.

As linhas serão organizadas por **Ano** e **Filme**.


```python
df = pd.DataFrame(todos_filmes)

df = df.sort_values(
    by=["Ano", "Filme"],
    ascending=[True, True]
).reset_index(drop=True)

display(df)
```

## Etapa 5 — Padronizar a coluna “Melhor Filme”

Os valores booleanos:

- `True` → **Sim**
- `False` → **Não**

serão convertidos para facilitar a leitura da tabela.


```python
df["Melhor Filme"] = df["Melhor Filme"].map({
    True: "Sim",
    False: "Não"
})

display(df)
```

## Etapa 6 — Verificar a quantidade de filmes por ano

Esta etapa é opcional, mas ajuda a conferir se os dados de todos os anos foram incluídos na tabela.


```python
resumo = (
    df.groupby("Ano")
      .size()
      .reset_index(name="Quantidade de filmes")
)

display(resumo)

print("Total geral:", len(df))
```

## Etapa 7 — Salvar os dados em Excel

Todos os anos serão armazenados em **uma única planilha**.


```python
nome_arquivo = "oscar_2010_2015.xlsx"

df.to_excel(
    nome_arquivo,
    index=False,
    engine="openpyxl"
)

print(f"Arquivo criado com sucesso: {nome_arquivo}")
```
